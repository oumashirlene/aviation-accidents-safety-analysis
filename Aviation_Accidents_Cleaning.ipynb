### Cleaning and constructing Key Measurables

Injuries and robustness to destruction are a key interest point for the client. Clean and impute relevant columns and then create derived fields that best quantifies what the client wishes to track. **Use commenting or markdown to explain any cleaning assumptions as well as any derived columns you create.**

**Construct metric for fatal/serious injuries**

*Hint:* Estimate the total number of passengers on each flight. The likelihood of serious / fatal injury can be estimated as a fraction from this.

```python
injury_columns = [
    "Total.Fatal.Injuries",
    "Total.Serious.Injuries",
    "Total.Minor.Injuries",
    "Total.Uninjured"
]

data[injury_columns].isnull().sum()
```

```python
# Convert injury columns to numeric
for column in injury_columns:
    data[column] = pd.to_numeric(data[column], errors="coerce")
```

```python
# Estimate the number of people aboard
data["Total.Occupants"] = data[injury_columns].sum(
    axis=1,
    min_count=1
)
```

```python
# Calculate total fatal and serious injuries
data["Fatal.Serious.Injuries"] = (
    data["Total.Fatal.Injuries"] +
    data["Total.Serious.Injuries"]
)
```

```python
# Calculate the fraction of occupants seriously or fatally injured
data["Injury.Rate"] = (
    data["Fatal.Serious.Injuries"] / data["Total.Occupants"]
)
```

```python
# Replace invalid results with missing values
data.loc[
    (data["Total.Occupants"] <= 0) |
    (data["Injury.Rate"] > 1),
    "Injury.Rate"
] = np.nan

data[[
    "Total.Occupants",
    "Fatal.Serious.Injuries",
    "Injury.Rate"
]].head()
```

**Aircraft.Damage**
- identify and execute any cleaning tasks
- construct a derived column tracking whether an aircraft was destroyed or not.

```python
# Inspect aircraft damage categories
data["Aircraft.damage"].value_counts(dropna=False)
```

```python
# Standardize damage descriptions
data["Aircraft.damage"] = (
    data["Aircraft.damage"]
    .astype("string")
    .str.strip()
    .str.lower()
)
```

```python
# Create a binary destruction indicator
data["Aircraft.Destroyed"] = np.where(
    data["Aircraft.damage"].fillna("").str.strip().str.lower() == "destroyed",
    1,
    np.where(data["Aircraft.damage"].isna(), np.nan, 0)
)

data[["Aircraft.damage", "Aircraft.Destroyed"]].head()
```

### Column Removal
- inspect the dataframe and drop any columns that have too many NaNs

```python
# Calculate percentage of missing values
missing_percent = data.isnull().mean() * 100
```

```python
# Display missing percentages
missing_percent.sort_values(ascending=False)
```

```python
# Protect essential columns from automatic removal
essential_columns = [
    "Event.Date",
    "Make",
    "Model",
    "Aircraft.Type",
    "Aircraft.Damage",
    "Aircraft.Destroyed",
    "Total.Occupants",
    "Fatal.Serious.Injuries",
    "Injury.Rate",
    "Engine.Type",
    "Weather.Condition",
    "Number.of.Engines",
    "Purpose.of.flight",
    "Broad.phase.of.flight"
]
```

```python
# Identify nonessential columns with over 70% missing data
columns_to_drop = [
    column for column in data.columns
    if missing_percent[column] > 70
    and column not in essential_columns
]

print("Columns to remove:", columns_to_drop)
```

```python
# Drop selected columns
data = data.drop(columns=columns_to_drop)
```
